## Ejercicio 3 – Listas y bucles

Crear una lista con 5 números\. Utilizando un bucle for, imprimir cada número de la lista\.

Luego:

Calcular la suma total de los números
Calcular el promedio
Mostrar ambos resultados por pantalla\.

```
lista_numeros=[20,5,85,100,40];
for i in lista_numeros:
    print(i);
```

```
#sumar los nùmeros de la lista
lista_numeros=[20,5,85,100,40];
for i in lista_numeros:
    suma=sum(lista_numeros);
print("La suma total es", suma);
```

```
# calcular el promedio de los nùmeros de la lista
lista_numeros=[20,5,85,100,40];
cont=0
for i in lista_numeros:
    suma=sum(lista_numeros);
    cont=cont+1;
    promedio=suma/cont;
print("El promedio es", promedio);
```

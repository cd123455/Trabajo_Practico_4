## Ejercicio 2 – Dataset simple con Pandas

```
lista=[["Ana",8],["Juan",6],["Carla",9],["Pedro",5],["Lucìa",7]]
```

```
df=pd.DataFrame(lista,columns=["Nombre","Nota"])
df
```

```
#promedio de las notas
promedio=df["Nota"].mean()
promedio
```

```
#Mostrar la nota máxima
NotaMaxima=df["Nota"].max()
NotaMaxima
```

```
#Mostrar la nota mìnima
NotaMinima=df["Nota"].min()
NotaMinima
```
